```python
# 1. Identify data type of given values.
number_1 = 100
text_1 = "Python"
number_2 = 45.67
text_2 = False
text_3 = [5, 10, 15]
text_4 = {"apple", "banana", "mango"}
text_5 = {"name": "Krish", "age": 20}

print(type(number_1).__name__)
print(type(text_1).__name__)
print(type(number_2).__name__)
print(type(text_2).__name__)
print(type(text_3).__name__)
print(type(text_4).__name__)
print(type(text_5).__name__)
```

```python
# 2. Convert int to float.
number = float(30)
print(number)
print(type(number).__name__)
```

```python
# 3. Convert float to int.
number = int(45.0)
print(number)
print(type(number).__name__)
```

```python
# 4. Create list of fruits.
fruits = ["Apple","Banana","Mango","Orange"]
print(fruits)
print(type(fruits).__name__)
```

```python
# 5. Create tuple of colors.
colors = ("Red","Blue","Green","Black")
print(colors)
print(type(colors).__name__)
```

```python
# 6. Create set with duplicates.
unique = {1,2,3,3,"Hello","Hello",5,4,1}
print(unique)
print(type(unique).__name__)
```

```python
# 7. Create dictionary of student.
student_data = {
    "name":"Anonymous",
    "roll_number" : 37,
    "standard" : 10 ,
    "result" : "pass"
}
print(student_data)
print(type(student_data).__name__)
```

```python
# 8. Difference list vs tuple.
list_data = [1,2,"name",5,6,True]
tuple_data = (1,2,"name",5,6,True)
print(list_data)
print(type(list_data).__name__)
print(tuple_data)
print(type(tuple_data).__name__)
# List --> Can be changed (mutable).
# Tuple --> Cannot be changed (immutable).
```

```python
# 9. Index a list.
list_data = [1,2,"name",5,6,True]
print(list_data[1])
print(list_data[-1])
print(list_data[4])
```

```python
# 10. Slice a string.
text = "Anonymous"
print(text[0:3])
print(text[-4:-1])
```

```python
# 11. Update dictionary.
student_data = {
    "name":"Anonymous",
    "roll_number" : 37,
    "standard" : 10 ,
    "result" : "pass"
}
student_data["attendance"] = "present"
print(student_data)
```

```python
# 12. Add/remove set elements.
unique = {1,2,3,3,"Hello","Hello",5,4,1}
unique.add(6)
unique.remove(1)
print(unique)
```

```python
# 13. Nested list.
list_data = [
    [10,40,"Hello"],
    [20,40,100],
    [45,65,70]
]
print(list_data)
print(list_data[0][2])
print(list_data[2][2])
print(type(list_data).__name__)
```

```python
# 14. Nested dictionary.
student_data ={
    "Student_1" : {
        "name" : "Anonymous",
        "roll_number" : 1,
        "age" : 21
    },
  "Student_2" : {
        "name" : "Anonymous1",
        "roll_number" : 2,
        "age" : 22
    }
}
print(student_data["Student_1"]["name"])
print(student_data["Student_2"]["name"])
```

```python
# 15. Mutable vs immutable example.
list_data = [1,2,"name",5,6,True]
tuple_data = (1,2,"name",5,6,True)

list_data.append(5)
print(list_data)
# Tuple is immutable, so it cannot be modified.
print(tuple_data)
```
