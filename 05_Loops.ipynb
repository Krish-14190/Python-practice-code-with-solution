## For Loop :

```python
# 1. Print numbers from 1 to 10.
for i in range(1,11):
  print(i)
```

```python
# 2. Print numbers from 10 to 1.
for i in range(10,0,-1):
  print(i)
```

```python
# 3. Print all even numbers from 1 to 100.
for i in range(1,101):
  if i % 2 == 0:
    print(i)
```

```python
# 4. Print all odd numbers from 1 to 100.
for i in range(1,101):
  if i % 2 != 0:
    print(i)
```

```python
# 5. Print the multiplication table of a given number.
n = int(input("Entr number : "))
for i in range(1,11):
  result = n*i
  print(n,"*",i," =",result)
```

```python
# 6. Find the sum of numbers from 1 to N.
n = int(input("Enter number : "))
total = 0
for i in range(1,n+1):
  total = total + i
print(total)
```

```python
# 7. Find the factorial of a number.
n = int(input("Enter a number : "))
fact = 1
for i in range(1,n+1):
  fact = fact * i
print(fact)
```

```python
# 8. Count the number of digits in an integer.
n = input("Enter digits : ")
count = 0
for i in n:
  count = count + 1
print(count)
```

```python
# 9. Check whether a number is a palindrome.
number = input("Enter a number: ")

reverse = ""

for digit in number:
    reverse = digit + reverse

if number == reverse:
    print("Palindrome")
else:
    print("Not a palindrome")
```

## While Loop :

```python
# 1. Print numbers from 1 to N using while.
number = int(input("Enter a number :  "))
i = 1
while i <= number :
  print(i)
  i = i+1
```

```python
# 2. Print the Fibonacci series up to N terms.

n = int(input("Enter number of terms: "))
a = 0
b = 1
count = 0

while count < n:
    print(a, end=" ")
    c = a + b
    a = b
    b = c
    count += 1
```
