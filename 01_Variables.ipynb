```python
# 15. Simple interest.
# formula for simple interest : SI = (PRT)/100
Principal = 10000
Rate = 5 # 5(%)
Time = 3 # 3 Year
si = (Principal * Rate * Time)/100
print(f"SI : {si:.2f}")
```

```python
# 16. Celsius to Fahrenheit.
# formula : Fahrenheit = (Celsius × 9 / 5) + 32
Celsius = 45
Fahrenheit = (Celsius * 9 /5 ) + 32
print(f"Fahrenheit : {Fahrenheit:.2f}")
```

```python
# 17. Hours to minutes and seconds.
# Minutes in an hour = 60
# Seconds in an hour = 3600

hour = 2
hour_to_minutes = hour * 60 # 60 minutes
hour_to_seconds = hour *3600 # 3600 seconds
print(f"{hour}h is {hour_to_minutes}m,\n{hour}h is {hour_to_seconds}s ")
```

```python
# 18. Compound interest.
# formula : CI = P × (1 + R/100)^T - P
Principal = 10000
Rate = 10 # 10(%)
Time = 2 # 2 Years
ci = Principal * (1 + Rate/100)** Time - Principal
print(f"CI : {ci:.2f}")
```
