## Input :

```python
# 1. Input and print name.
name = input("Enter your name : ")
print(name)
```

```python
# 2. Input age.
age = int(input("Enter your age : "))
print(age)
```

```python
# 3. Add two input numbers.
number_1 = int(input("Enter number : "))
number_2 = int(input("Enter number : "))
sum = number_1 + number_2
print(sum)
```

```python
# 4. Area of rectangle from input.
#Formula : A = wl
width = int(input("Enter width : "))
length = int(input("Enter length : "))
area = width * length
print(area)
```

```python
# 5. Swap two input values.
a = int(input("Enter a value : "))
b = int(input("Enter a value : "))
a,b=b,a
print(a)
print(b)
```

```python
# 6. Input marks and average.
marks_1 = int(input("Enter marks : "))
marks_2 = int(input("Enter marks : "))
marks_3 = int(input("Enter marks : "))
avg = (marks_1+marks_2+marks_3)/3
print(avg)
```

## Operator :

```python
# 1. Addition.
a = 10
b = 20
add = a+b
print(add)
```

```python
# 2. Subtraction.
a = 20
b = 10
sub = a-b
print(sub)
```

```python
# 3. Multiplication.
a = 10
b = 10
mul = a * b
print(mul)
```

```python
# 4. Division.
a = 10
b = 5
div = a / b
print(div)
```

```python
# 5. Floor division.
a = 10
b = 3
fd = a // b
print(fd)
```

```python
# 6. Modulus.
a = 10
b = 3
mod = a % b
print(mod)
```

```python
# 7. Exponent.
a = 4
b = 2
exp = a ** 2
print(exp)
```

```python
# 8. Compare two numbers.
a = 10
b = 8
print(a>b)
print(a<b)
```

```python
# 9. Logical AND.
a = 10
b = 8

print(a > b and a == 10)
```

```python
# 10. Logical OR.
a = 10
b = 8

print(a > b or a < b)
```

```python
# 11. Logical NOT.
a = 10
b = 8

print(not(a > b))
```
